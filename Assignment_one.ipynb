Data validation
Let us initialize a variable correct_answers to keep track of the number of correct answers.

Iterate through the states_capitals array, prompting the user to enter the capital for each state.

Convert the user's answer to lowercase to make the comparison case-insensitive.

Compare the user's answer with the correct capital (also converted to lowercase) and increment the correct_answers variable if the answer is correct.

If the answer is incorrect, print the correct capital to provide feedback to the user.

```python
# Print the total number of states
print("Total number of states:", len(states_capitals))

# Verify that there are exactly 50 states
if len(states_capitals) == 50:
    print("Data validation: There are exactly 50 state-capital pairs.")
else:
    print("Data validation error: There should be exactly 50 states.")

# Print the first 5 states alphabetically
print("\nFirst 5 states and capitals:")

for state in sorted(states_capitals)[:5]:
    print(f"{state}: {states_capitals[state]}")
    # ---------------------------------------------------------
# Part 2: Quiz Implementation
# ---------------------------------------------------------
# Welcome message
print("=" * 50)
print("Welcome to the U.S. States and Capitals Quiz!")
print("Test your knowledge by answering 5 questions.")
print("Good luck!")
print("=" * 50)
print("\n" + "=" * 50)
print("       U.S. STATES AND CAPITALS QUIZ")
print("=" * 50)

import random
print("\nYou will be asked 5 questions.")
print("Enter the capital city for each state.")

# Randomly select 5 different states
selected_states = random.sample(list(states_capitals.keys()), 5)

# Keep track of the score
score = 0

# Quiz loop
for question_number, state in enumerate(selected_states, start=1):

    print(f"\n[{question_number}] What is the capital of {state}?")

    # Input validation for empty responses
    while True:
        user_answer = input("Your answer: ").strip()

        if user_answer:
            break

        print("Please enter an answer.")

    # Case-insensitive comparison
    if user_answer.lower() == states_capitals[state].lower():
        print("Correct!")
        score += 1
    else:
        print(
            f"Incorrect. The correct answer is "
            f"{states_capitals[state]}."
        )
# ---------------------------------------------------------
# Final Score
# ---------------------------------------------------------

print("\n" + "=" * 50)
print(f"Your score is {score} out of 5")
print("=" * 50)
```
